# Import 

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from pandas.plotting import autocorrelation_plot
from statsmodels.tsa.stattools import adfuller
```

# Load data sets


```python
climate_df = pd.read_csv('data/02-processed/climate_clean.csv')
deflated_df = pd.read_csv('data/02-processed/deflated-prices.csv')
```

```python
climate_df.head()
```

# Convert Climate to DatetimeIndex

```python
# Convert column to datetime
climate_df['year_month'] = pd.to_datetime(climate_df['year_month'])

# Set it as index
climate_df = climate_df.set_index('year_month')

# Rename index
climate_df.index.name = 'date'

# Sort
climate_df = climate_df.sort_index()

climate_df.head()
```

```python
climate_df.head()
```

```python
deflated_df.head()
```

# Convert deflated to datetimeindex

```python
# Convert date column to datetime
deflated_df['date'] = pd.to_datetime(deflated_df['date'])

# Set as index
deflated_df = deflated_df.set_index('date')

# Sort just to be safe
deflated_df = deflated_df.sort_index()

deflated_df.head()
```

# Merge the two datasets


```python
merged_df = climate_df.join(deflated_df, how='inner')
```
