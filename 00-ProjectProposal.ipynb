## Research Question

To what extent do climate anomalies in the major coffee-exporting countries of Brazil and Vietnam, energy market dynamics via oil prices, and U.S. oil production, and U.S. inflation jointly influence variation and volatility in global coffee prices between 2009–2021 on a daily basis?

## Background and Prior Work

#### Introduction:

Sitting down with a cup of coffee at one of the many coffee shops at UC San Diego on a cool day during winter quarter doesn’t just burn your tongue; it also puts a major dent in your wallet. With the price of a latte on campus ranging from 4.25 at Middle of Muir to 7.75 at Bird Rock Roasters, it is clear that a cup of coffee is not a trivial expense, yet many are willing to pay for it. 

Global coffee prices have seen a significant increase over the past decades from 2009 to 2021 with key elements of uncertainty shifting based on supply-demand imbalances of bean production, varying climate circumstances, and macroeconomic conditions of inflation. With coffee being one of the most traded commodities in the world, understanding the factors driving the price fluctuations—and whether there exists factors exhibiting predictable lag effects—allows understanding for major coffee-exporting countries of Brazil and Vietnam. 

Prior research has used autoregressive distributed lag (ARDL) and vector autoregression (VAR) models to analyze how climate shocks and energy prices affect agricultural commodity markets, finding significant delayed effects between weather anomalies and crop prices.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5)   

#### Supply & Consumer Prices:

Supply prices are, of course, inextricably intertwined with what those who demand are willing to pay. Looking at the broader San Diego area, a poll conducted by Talitha Coffee Roasters determined that 97% of their patrons would be willing to pay up to 8 dollars for a cup of coffee, with an upper bound of 12 dollars.<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7) One San Diego coffee shop chain, Bird Rock Coffee Roasters, shows an increase of 6 to 7% in retail prices year over year.<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7) Many business owners cite the rising prices of coffee beans as the source of the augmented consumer cost, as raw bean prices account for a minimum of 40% of the total cost of producing a bag of coffee.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Brazil, which is responsible for over a third of global coffee production, has been affected by extreme droughts due to climate shocks. The droughts of 2021 and 2022 led to one of the worst harvests seen in 20 years, and 2025 has continued that trend with a new drought.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) The lack of supply has been speculated as a main source of rising prices, but there are other factors at play as well. 

#### Inflation Rate & Coffee Production:

Inflation has also been cited as a potential source of rising prices, as grocery prices have been increasing overall. The Bureau of Labor Statistics’ Consumer Price Index showed that the index for food and energy rose 2.6% between 2024 and 2025.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) Coffee, however, has spiked far above the 2.9% rate of inflation at a 20% increase, alongside many other household staples like eggs (10%), beef (14%), bananas (6%), and bacon (5%).<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) Direct and indirect energy costs almost certainly play a role in rising prices, as the cost of oil and natural gas can account for nearly half of the total costs for food production, of which coffee is categorically inclined.<a name="cite_ref-8"></a>[<sup>8</sup>](#cite_note-8) There are, of course, myriad other factors that may contribute to these rising costs, but varying climate, inflation, and energy costs seem to be at the core of the increase. Each of these factors operates on a vastly different time scale from the others. Inflation is a gradual incline over many years, generally lacking extreme fluctuations. Climate, however, is highly variable and displays many inflection points depending on region and time of year. Oil prices are even more volatile, with near-immediate downstream effects on production costs.
