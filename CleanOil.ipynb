```python
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

```python
df = pd.read_csv('/Users/nischalpanthi/Group048_WI26/data/00-raw/crude-oil-price.csv')
df.head(5)
```

```python
df.isnull().sum()
df_clean = df.dropna()
df_clean.isnull().sum()
df_clean.head(5)
```

```python
df_clean = df.dropna().copy()
df_clean['date'] = pd.to_datetime(df_clean['date'])
```

```python
num_duplicates = df_clean.duplicated(subset=['date']).sum()
#print("Duplicates after:", num_duplicates)
df_clean = df_clean.drop_duplicates(subset=['date']).reset_index(drop=True)
```

```python
df_clean['price'] = pd.to_numeric(df_clean['price'], errors='coerce')
df_clean = df_clean.sort_values('date').reset_index(drop=True)
```

```python
print("Shape:", df_clean.shape)
print("Date range:", df_clean['date'].min(), "to", df_clean['date'].max())
print("Date gaps (days):")
print(df_clean['date'].diff().dt.days.describe())
```
