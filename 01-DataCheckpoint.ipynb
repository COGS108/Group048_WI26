## Hypothesis


If these are indeed the most significant factors, this begs the question of whether these can be used as predictors for increases in coffee prices. We suspect that adverse climate anomalies in major coffee-producing regions, specifically Brazil and Vietnam, will be positively associated with increases in global coffee prices at lags of approximately 6–12 months. This expectation is grounded in the production cycle of coffee, where climate shocks such as drought, excessive rainfall, or heat stress disrupt flowering and bean development. This leads to reduced yields that are realized during subsequent harvest cycles. Because coffee production and export processes take place over several months, supply shortfalls are expected to manifest in market prices with a delayed effect. Thus, negative climate shocks (e.g., unusually low precipitation or extreme temperature anomalies) are expected to increase coffee prices following a 6-12 month lag consistent with agricultural timing.
With regard to the effects of oil, we hypothesize that increases in global oil prices will be positively associated with real coffee prices at shorter lags of approximately 0–3 months. Oil prices affect coffee markets primarily through cost transmission. Rising energy prices increase transportation, fertilizer, and processing costs. These mechanisms operate more swiftly than agricultural supply adjustments, suggesting that oil price shocks should positively affect coffee prices within weeks to a few months.
Because climate and oil prices influence coffee markets through distinct transmission mechanisms, one operating through delayed physical supply effects and the other through immediate cost, we expect their effects to appear at different time points within a unified multivariate framework. ### COFFEE PRICES by DATE

This is the core dataset that encompasses the daily price of coffee (USD per pound) from Aug 1973. The variables here are date and price per pound Data is from Macrotrends.net, a website that specializes in market analysis Last updated Dec 2023. This dataset contains historical daily coffee prices beginning in 1973. It consists of two variables: date and value. The date column records the trading day (in YYYY-MM-DD format), and each row represents a single observation in a daily time series. The value column represents the market price of coffee futures on that specific date. Prices are expressed in U.S. dollars per pound (USD/lb).

For example, a value of 0.6735 corresponds to $0.6735 per pound, or approximately 67.35 cents per pound. Coffee is a globally traded agricultural commodity, and these prices reflect the benchmark futures market price rather than retail grocery store prices. Historically, coffee prices have fluctuated substantially due to weather events (particularly in Brazil, the world’s largest producer), oil prices affecting transportation costs, global inflation, currency exchange rates, and supply chain disruptions. Sustained increases in price may indicate tightening global supply or macroeconomic pressures, while prolonged declines may reflect overproduction or weak demand.

Because this dataset is a financial time series, it allows for analysis of long-term trends, short-term volatility, price cycles, and structural breaks over decades. However, there are important limitations. First, the dataset contains only price and date, with no explanatory variables such as weather, production levels, exchange rates, or oil prices. This limits causal analysis unless additional external data is merged. Second, the prices are nominal U.S. dollars and are not adjusted for inflation, meaning long-term comparisons may overstate real price growth. Third, futures prices are influenced not only by physical supply and demand but also by speculation and hedging activity in financial markets, which can introduce volatility unrelated to immediate production conditions.

Overall, this dataset provides a clean and long historical record of coffee futures prices, making it useful for time-series analysis, volatility modeling, and macroeconomic correlation studies. However, deeper economic explanations require integration with additional environmental and financial data sources.

# Outliners

```python
coffee_df['value'].describe()
```

```python
sns.boxplot(x=coffee_df['value'])
```

```python
Q1 = coffee_df['value'].quantile(0.25)
Q3 = coffee_df['value'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR


outliers = coffee_df[(coffee_df['value'] < lower) | (coffee_df['value'] > upper)]

print(outliers)
```

Using the 1.5×IQR rule, 461 observations were flagged as outliers. These values correspond to historically documented periods of extreme coffee price volatility, including major supply shocks in the late 1970s and early 2010s. Because coffee prices are known to exhibit cyclical commodity spikes, these observations reflect real economic events rather than data entry errors. Therefore, they were retained in the dataset.

```python
# freq='B' indicating "Buisness Days" 
full_range = pd.date_range(
    start=nonull_coffee_df['date'].min(),
    end=nonull_coffee_df['date'].max(),
    freq='B'
)

full_range
```

```python
nonull_coffee_df = nonull_coffee_df.sort_values('date')

nonull_coffee_df['date'] = pd.to_datetime(nonull_coffee_df['date'])

nonull_coffee_df['date'].is_monotonic_increasing
```

# Find total number of missing days

```python
actual_dates = pd.DatetimeIndex(nonull_coffee_df['date'])

missing_days = full_range.difference(actual_dates)

print("Missing days:",len(missing_days))
```

Approximately 493 business days (≈3.8% of the expected trading days) were not present in the dataset. Given the 50-year historical range, these gaps likely reflect exchange-specific holiday schedules, historical trading suspensions, or data source limitations rather than data corruption. No large contiguous blocks of missing dates were detected.
