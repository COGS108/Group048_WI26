Kadhim used autoregressive distributed lag (ARDL) and vector autoregression (VAR) models to analyze how climate shocks and energy prices affect agricultural commodity markets with a focal point on tea but also included coffee. The ARDL process analyzes both short term and long term dynamics within the data. It also allows one to determine correlations between a stationary and non-stationary variable without having to test the stationarity of all variables prior to estimating the model. While we liked the concept of using a lagged relationship due to the delayed transmission of cost differences and the modeling based on individual mechanisms (climate and oil, we ultimately decided to go with a multivariate single-equation regression with lags. In VAR, all of the variables depend on each other, which we did not think fit the problem, as there seems to be a clear directional structure between affecting factors and coffee. Using ARDL/VAR would also increase our parameter count, thereby increasing our risk of overfitting without a substantial benefit, so we believe that our approach, in its simplicity, will give us a better mechanistic understanding of how the price of coffee is affected.

In addition to the Kadhim study, we also explored established and public records outlining the calculations on how to adjust prices of coffee throughout the years for inflation to properly compare the data.<a name="cite_ref-9"></a>[<sup>9</sup>](#cite_note-9) The Consumer Price Index (CPI) measures the change in prices of any goods or services paid by consumers. The amount of each consumable is kept constant (deemed “baskets”), allowing CPI to be calculated using the formula
$
\text{CPI}_t = \left( \frac{\text{Cost of Basket in Current Period}}{\text{Cost of Basket in Base Period}} \right) \times 100
$
which can then be used to calculate the inflation rate via
$
\text{Inflation Rate}_t = \left( \frac{\text{New CPI} - \text{Prior CPI}}{\text{Prior CPI}} \right) \times 100
$. 
This gave us the idea to deflate the oil and coffee prices for inflation so that we could remove that as a factor driving increases in coffee prices. We hope that this will grant a more accurate view of the actual effects of our factors of interest.
Finally, previous work by the International Coffee Organization has documented a positive correlation between crude oil prices and coffee futures prices between 1990-2014.<a name="cite_ref-10"></a>[<sup>10</sup>](#cite_note-10) This confirmed that we were on the right path by looking at oil prices, but we aim to build upon this in several ways. The regression in this case was contemporaneous, which does not line up with our expectation of a delayed effect. They also did not deflate for inflation, using nominal futures prices as their metric for coffee prices. This means that the rise of coffee prices seen in their result could be in part due to inflation, which is a confound we would prefer to remove from our analysis.

We suspect that adverse climate anomalies in major coffee-producing regions, specifically Brazil and Vietnam, will be positively associated with increases in global coffee prices at lags of approximately 6–12 months. This expectation is grounded in the production cycle of coffee, where climate shocks such as drought, excessive rainfall, or heat stress disrupt flowering and bean development. This leads to reduced yields that are realized during subsequent harvest cycles. Because coffee production and export processes take place over several months, supply shortfalls are expected to manifest in market prices with a delayed effect. Thus, negative climate shocks (e.g., unusually low precipitation or extreme temperature anomalies) are expected to increase coffee prices following a 6-12 month lag consistent with agricultural timing.
With regard to the effects of oil, we hypothesize that increases in global oil prices will be positively associated with real coffee prices at shorter lags of approximately 0–3 months. Oil prices affect coffee markets primarily through cost transmission. Rising energy prices increase transportation, fertilizer, and processing costs. These mechanisms operate more swiftly than agricultural supply adjustments, suggesting that oil price shocks should positively affect coffee prices within weeks to a few months.
Because climate and oil prices influence coffee markets through distinct transmission mechanisms, one operating through delayed physical supply effects and the other through immediate cost, we expect their effects to appear at different time points within a unified multivariate framework. ### COFFEE PRICES by DATE

This is the core dataset that encompasses the daily price of coffee (USD per pound) from Aug 1973. The variables here are date and price per pound Data is from Macrotrends.net, a website that specializes in market analysis Last updated Dec 2023. This dataset contains historical daily coffee prices beginning in 1973. It consists of two variables: date and value. The date column records the trading day (in YYYY-MM-DD format), and each row represents a single observation in a daily time series. The value column represents the market price of coffee futures on that specific date. Prices are expressed in U.S. dollars per pound (USD/lb).

For example, a value of 0.6735 corresponds to $0.6735 per pound, or approximately 67.35 cents per pound. Coffee is a globally traded agricultural commodity, and these prices reflect the benchmark futures market price rather than retail grocery store prices. Historically, coffee prices have fluctuated substantially due to weather events (particularly in Brazil, the world’s largest producer), oil prices affecting transportation costs, global inflation, currency exchange rates, and supply chain disruptions. Sustained increases in price may indicate tightening global supply or macroeconomic pressures, while prolonged declines may reflect overproduction or weak demand.

Because this dataset is a financial time series, it allows for analysis of long-term trends, short-term volatility, price cycles, and structural breaks over decades. However, there are important limitations. First, the dataset contains only price and date, with no explanatory variables such as weather, production levels, exchange rates, or oil prices. This limits causal analysis unless additional external data is merged. Second, the prices are nominal U.S. dollars and are not adjusted for inflation, meaning long-term comparisons may overstate real price growth. Third, futures prices are influenced not only by physical supply and demand but also by speculation and hedging activity in financial markets, which can introduce volatility unrelated to immediate production conditions.

Overall, this dataset provides a clean and long historical record of coffee futures prices, making it useful for time-series analysis, volatility modeling, and macroeconomic correlation studies. However, deeper economic explanations require integration with additional environmental and financial data sources.

## Crude Oil Monthly Spot Prices (WTI, 1983–2026)

This dataset contains monthly West Texas Intermediate (WTI) crude oil spot prices 
sourced from macrotrends.net, spanning from March 1983 to February 2026 — over 
40 years of data. It has 4 columns:

- **`date`**: The first day of each month, representing that month's average price observation.
- **`price`**: The average WTI crude oil spot price for that month, measured in **USD per barrel**. 
  Historically, prices have ranged from roughly **10 dollar per barrel** (1998 OPEC oversupply glut) to ~ **134 dollar per barrel**
  (June 2008, pre-financial crisis peak). Prices below **30 dollar per barrel** are generally considered very low 
  and are associated with demand shocks or oversupply; prices above **100 dollar per barrel** have historically 
  triggered economic slowdowns due to high energy costs.
- **`change`**: The absolute month-over-month change in price, in **USD per barrel** 
- **`percentChange`**: The month-over-month change expressed as a **percentage** of the prior 
  month's price. Both `change` and `percentChange` are 
  derived directly from `price`, so they are redundant for modeling but useful for EDA.

This is a macroeconomic time-series dataset with no sampling bias in the traditional sense:
WTI prices are set by open commodity markets and recorded objectively. However, there are 
notable concerns. The data is aggregated to monthly averages, which hides extreme 
intra-month volatility; for example, WTI briefly went negative in April 2020 during COVID-19, but the monthly average that month was around $16.55/barrel, 
obscuring that historic anomaly. Also, this dataset reflects only one benchmark crude grade 
(WTI)Other grades like Brent crude or Dubai crude trade at different prices, so this data 
may not fully represent the global oil market.

```python
df_clean = df_oil.dropna(how='any').copy()
df_clean['date'] = pd.to_datetime(df_clean['date'])
df_clean['price'] = pd.to_numeric(df_clean['price'], errors='coerce')

print("After dropna")
print("Missing values:", df_clean.isnull().sum().sum())
print("Shape:", df_clean.shape)
df_clean.head()
```

## Duplicates & Date Ordering

```python
num_duplicates = df_clean.duplicated(subset='date').sum()
print(f"Duplicate date rows: {num_duplicates}")

df_clean = df_clean.drop_duplicates(subset='date').sort_values('date').reset_index(drop=True)
print("Date range:", df_clean['date'].min(), "to", df_clean['date'].max())
print("Date gaps (days):")
print(df_clean['date'].diff().dt.days.describe())
```

## Outlier Detection

```python
Q1 = df_clean['price'].quantile(0.25)
Q3 = df_clean['price'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df_clean[(df_clean['price'] < lower) | (df_clean['price'] > upper)]
print(f"IQR range: [{lower:.2f}, {upper:.2f}]")
print(f"\n{len(outliers)} outlier months flagged:\n")
print(outliers[['date', 'price', 'percentChange']].to_string())

# Boxplot
plt.figure(figsize=(6, 4))
plt.boxplot(df_clean['price'])
plt.title('WTI Crude Oil Price')
plt.ylabel('Price (USD per barrel)')
plt.show()
```

The IQR method flags the high-price episodes driven by the 2007–2008 commodity supercycle 
(peak: **134 dollar per barrel** in June 2008) and the 2022 Russia-Ukraine war (**123 dollar per barrel** in March 2022), 
as well as extreme low-price months during the 2020 COVID-19 demand collapse (**16–17dollar per barrel**). 
These are real, historically documented events — not data entry errors. We retain all 
outliers in the cleaned dataset, as they represent genuine market extremes that are meaningful 
for analysis. We flag them with an is_outlier column for optional use downstream.

## Flag Outliers

```python
df_clean['is_outlier'] = (df_clean['price'] < lower) | (df_clean['price'] > upper)
print(f"Outlier flag added. {df_clean['is_outlier'].sum()} rows flagged.")
df_clean.head()
```
