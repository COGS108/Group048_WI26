Kadhim used autoregressive distributed lag (ARDL) and vector autoregression (VAR) models to analyze how climate shocks and energy prices affect agricultural commodity markets with a focal point on tea but also included coffee. The ARDL process analyzes both short term and long term dynamics within the data. It also allows one to determine correlations between a stationary and non-stationary variable without having to test the stationarity of all variables prior to estimating the model. While we liked the concept of using a lagged relationship due to the delayed transmission of cost differences and the modeling based on individual mechanisms (climate and oil, we ultimately decided to go with a multivariate single-equation regression with lags. In VAR, all of the variables depend on each other, which we did not think fit the problem, as there seems to be a clear directional structure between affecting factors and coffee. Using ARDL/VAR would also increase our parameter count, thereby increasing our risk of overfitting without a substantial benefit, so we believe that our approach, in its simplicity, will give us a better mechanistic understanding of how the price of coffee is affected.

In addition to the Kadhim study, we also explored established and public records outlining the calculations on how to adjust prices of coffee throughout the years for inflation to properly compare the data.<a name="cite_ref-9"></a>[<sup>9</sup>](#cite_note-9) The Consumer Price Index (CPI) measures the change in prices of any goods or services paid by consumers. The amount of each consumable is kept constant (deemed “baskets”), allowing CPI to be calculated using the formula
$
\text{CPI}_t = \left( \frac{\text{Cost of Basket in Current Period}}{\text{Cost of Basket in Base Period}} \right) \times 100
$
which can then be used to calculate the inflation rate via
$
\text{Inflation Rate}_t = \left( \frac{\text{New CPI} - \text{Prior CPI}}{\text{Prior CPI}} \right) \times 100
$. 
This gave us the idea to deflate the oil and coffee prices for inflation so that we could remove that as a factor driving increases in coffee prices. We hope that this will grant a more accurate view of the actual effects of our factors of interest.
Finally, previous work by the International Coffee Organization has documented a positive correlation between crude oil prices and coffee futures prices between 1990-2014.<a name="cite_ref-10"></a>[<sup>10</sup>](#cite_note-10) This confirmed that we were on the right path by looking at oil prices, but we aim to build upon this in several ways. The regression in this case was contemporaneous, which does not line up with our expectation of a delayed effect. They also did not deflate for inflation, using nominal futures prices as their metric for coffee prices. This means that the rise of coffee prices seen in their result could be in part due to inflation, which is a confound we would prefer to remove from our analysis.

# Hypothesis


If these are indeed the most significant factors, this begs the question of whether these can be used as predictors for increases in coffee prices. We suspect that adverse climate anomalies in major coffee-producing regions, specifically Brazil and Vietnam, will be positively associated with increases in global coffee prices at lags of approximately 6–12 months. This expectation is grounded in the production cycle of coffee, where climate shocks such as drought, excessive rainfall, or heat stress disrupt flowering and bean development. This leads to reduced yields that are realized during subsequent harvest cycles. Because coffee production and export processes take place over several months, supply shortfalls are expected to manifest in market prices with a delayed effect. Thus, negative climate shocks (e.g., unusually low precipitation or extreme temperature anomalies) are expected to increase coffee prices following a 6-12 month lag consistent with agricultural timing.
With regard to the effects of oil, we hypothesize that increases in global oil prices will be positively associated with real coffee prices at shorter lags of approximately 0–3 months. Oil prices affect coffee markets primarily through cost transmission. Rising energy prices increase transportation, fertilizer, and processing costs. These mechanisms operate more swiftly than agricultural supply adjustments, suggesting that oil price shocks should positively affect coffee prices within weeks to a few months.
Because climate and oil prices influence coffee markets through distinct transmission mechanisms, one operating through delayed physical supply effects and the other through immediate cost, we expect their effects to appear at different time points within a unified multivariate framework. We further suspect there may be outliers in the data resulting from geopolitical/macroeconomic events not covered within the scope of our analysis.

This is the core dataset that encompasses the daily price of coffee (USD per pound) from Aug 1973. The variables here are date and price per pound Data is from Macrotrends.net, a website that specializes in market analysis Last updated Dec 2023. This dataset contains historical daily coffee prices beginning in 1973. It consists of two variables: date and value. The date column records the trading day (in YYYY-MM-DD format), and each row represents a single observation in a daily time series. The value column represents the market price of coffee futures on that specific date. Prices are expressed in U.S. dollars per pound (USD/lb).

For example, a value of 0.6735 corresponds to $0.6735 per pound, or approximately 67.35 cents per pound. Coffee is a globally traded agricultural commodity, and these prices reflect the benchmark futures market price rather than retail grocery store prices. Historically, coffee prices have fluctuated substantially due to weather events (particularly in Brazil, the world’s largest producer), oil prices affecting transportation costs, global inflation, currency exchange rates, and supply chain disruptions. Sustained increases in price may indicate tightening global supply or macroeconomic pressures, while prolonged declines may reflect overproduction or weak demand.

Because this dataset is a financial time series, it allows for analysis of long-term trends, short-term volatility, price cycles, and structural breaks over decades. However, there are important limitations. First, the dataset contains only price and date, with no explanatory variables such as weather, production levels, exchange rates, or oil prices. This limits causal analysis unless additional external data is merged. Second, the prices are nominal U.S. dollars and are not adjusted for inflation, meaning long-term comparisons may overstate real price growth. Third, futures prices are influenced not only by physical supply and demand but also by speculation and hedging activity in financial markets, which can introduce volatility unrelated to immediate production conditions.

Overall, this dataset provides a clean and long historical record of coffee futures prices, making it useful for time-series analysis, volatility modeling, and macroeconomic correlation studies. However, deeper economic explanations require integration with additional environmental and financial data sources.

### INFLATION (CPI) MONTH-DATE

https://data.bls.gov/timeseries/CUUR0000SA0

This dataset is the Consumer Price Index for All Urban Consumers (CPI-U), taken directly from the U.S. Bureau of Labor Statistics. The CPI-U measures price changes for goods and services like food, energy, and shelter for over 90% of the U.S. population. The population is comprised of people from a diverse range of backgrounds and careers, as well as a diverse geographical sampling (75 urban areas and 23,000 retail establishments). The CPI-U is an index, with the values representing the cost of a fixed "basket" of the aforementioned goods and services across the population. Values range from as low as 9.7 to as high as 324.8, with these values increasing over time. The only important metric in this dataset is the CPI index level, as all the dataset contains is the index level and the date it was collected.

There are a couple of limitations to using CPI as our measurement. While CPI is a standard measure of consumer price inflation, it reflects U.S. consumer prices rather than global commodity production costs. By using CPI to adjust coffee prices, we are expressing prices in constant U.S. purchasing-power terms rather than producer-cost terms. As well, the methods for calculating CPI have changed over time (updates to the consumption basket, weighting procedures, quality adjustments, and data collection methods), which may slightly affect long-run comparisons. However, CPI is still the standard measure used to adjust for inflation in economic research and is appropriate for converting nominal prices into real terms for this analysis.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

cpi_df = pd.read_csv("https://www.dropbox.com/scl/fi/kdm18c2281w9bb8p0a73k/inflation_CPI.csv?rlkey=xin8u602b888dhvy70194zpz8&st=ay9g1uyr&dl=1")

print(cpi_df.shape)

print(cpi_df.head(10))
print(cpi_df.tail(10))

#This is not tidy OR clean :(
#The months are column headers (except they're just in the first row and not actually headers), and the years are row names. This is NOT TIDY
#There appears to be at least one entry (Oct 2025) that has no value. This is NOT CLEAN
```

```python
#First row has real column names; make that the colnames and remove from data
cpi_df.columns = cpi_df.iloc[0].values
cpi_df = cpi_df.drop(cpi_df.index[0]).reset_index(drop=True)

#Take the data from hamburger style to hot dog style so the timeseries will work; months to rows, cols month and cpi
cpi_long = cpi_df.melt(id_vars="Year",
    value_vars=["Jan", "Feb", "Mar", "Apr", "May", "June",
                "July", "Aug", "Sep", "Oct", "Nov", "Dec"],
    var_name="month",
    value_name="cpi")


#Convert CPI to numeric because it had some weird values instead of NaN like a normal dataset
cpi_long["cpi"] = pd.to_numeric(cpi_long["cpi"], errors="coerce")

#Drop any NaN values
cpi_long = cpi_long.dropna(subset=["cpi"])


#Map month names to numbers to eventually build dates
month_map = {
    "Jan": 1, "Feb": 2, "Mar": 3, "Apr": 4,
    "May": 5, "June": 6, "July": 7, "Aug": 8,
    "Sep": 9, "Oct": 10, "Nov": 11, "Dec": 12
}

cpi_long["month_num"] = cpi_long["month"].map(month_map)

#Build datetime that matches the formatting of the other datasets
cpi_long["date"] = pd.to_datetime(
    dict(
        year=cpi_long["Year"],
        month=cpi_long["month_num"],
        day=1
    )
)


#Make the monthly CPI table and sort chronologically
cpi_monthly = (
    cpi_long.dropna(subset=["cpi"])
    .sort_values("date")[["date", "cpi"]]
    .set_index("date")
)


#Make sure no sneaky strings snuck through
cpi_monthly['cpi'] = pd.to_numeric(cpi_monthly['cpi'], errors='coerce')
cpi_monthly = cpi_monthly.dropna(subset=['cpi'])

cpi_monthly = cpi_monthly.sort_index()

# Daily date index so that this includes all possible dates, can pare this down to our actual date range when merging datasets
start = cpi_monthly.index.min()
end = cpi_monthly.index.max() + pd.offsets.MonthEnd(0)

#Full calendar by day
daily_index = pd.date_range(start=start, end=end, freq="D")

cpi_daily = pd.DataFrame(index=daily_index)

# Map each day to its month, then merge CPI by month
cpi_daily["year_month"] = cpi_daily.index.to_period("M")

cpi_m = cpi_monthly.copy()
cpi_m["year_month"] = cpi_m.index.to_period("M")


#No data for October 2025, explicitly dropping that month because it isn't in the date range we're going to use anyway
missing_month = pd.Period("2025-10", freq="M")

cpi_monthly = cpi_monthly[
    cpi_monthly.index.to_period("M") != missing_month
]

print(cpi_monthly)
print(cpi_monthly.columns)
```

### Crude Oil Monthly Spot Prices (WTI, 1983–2026)

This dataset contains monthly West Texas Intermediate (WTI) crude oil spot prices 
sourced from macrotrends.net, spanning from March 1983 to February 2026 — over 
40 years of data. It has 4 columns:

- **`date`**: The first day of each month, representing that month's average price observation.
- **`price`**: The average WTI crude oil spot price for that month, measured in **USD per barrel**. 
  Historically, prices have ranged from roughly **10 dollar per barrel** (1998 OPEC oversupply glut) to ~ **134 dollar per barrel**
  (June 2008, pre-financial crisis peak). Prices below **30 dollar per barrel** are generally considered very low 
  and are associated with demand shocks or oversupply; prices above **100 dollar per barrel** have historically 
  triggered economic slowdowns due to high energy costs.
- **`change`**: The absolute month-over-month change in price, in **USD per barrel** 
- **`percentChange`**: The month-over-month change expressed as a **percentage** of the prior 
  month's price. Both `change` and `percentChange` are 
  derived directly from `price`, so they are redundant for modeling but useful for EDA.

This is a macroeconomic time-series dataset with no sampling bias in the traditional sense:
WTI prices are set by open commodity markets and recorded objectively. However, there are 
notable concerns. The data is aggregated to monthly averages, which hides extreme 
intra-month volatility; for example, WTI briefly went negative in April 2020 during COVID-19, but the monthly average that month was around $16.55/barrel, 
obscuring that historic anomaly. Also, this dataset reflects only one benchmark crude grade 
(WTI)Other grades like Brent crude or Dubai crude trade at different prices, so this data 
may not fully represent the global oil market.

### Load Dataset

```python
# Tidy data: each row = one monthly observation, each column = one variable
# Verify: no multi-value cells, no variables stored as row headers
print("Columns:", df_oil.columns.tolist())
print("Shape:", df_oil.shape)
print("\nEach row is one month's observation — data is already tidy.")
df_oil.head()
```

### Clean: Drop NaN

```python
df_clean = df_oil.dropna(how='any').copy()
df_clean['date'] = pd.to_datetime(df_clean['date'])
df_clean['price'] = pd.to_numeric(df_clean['price'], errors='coerce')

print("After dropna")
print("Missing values:", df_clean.isnull().sum().sum())
print("Shape:", df_clean.shape)
df_clean.head()
```

### Outlier Detection

```python
Q1 = df_clean['price'].quantile(0.25)
Q3 = df_clean['price'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df_clean[(df_clean['price'] < lower) | (df_clean['price'] > upper)]
print(f"IQR range: [{lower:.2f}, {upper:.2f}]")
print(f"\n{len(outliers)} outlier months flagged:\n")
print(outliers[['date', 'price', 'percentChange']].to_string())

# Boxplot
plt.figure(figsize=(6, 4))
plt.boxplot(df_clean['price'])
plt.title('WTI Crude Oil Price')
plt.ylabel('Price (USD per barrel)')
plt.show()
```

The IQR method flags the high-price episodes driven by the 2007–2008 commodity supercycle 
(peak: **134 dollar per barrel** in June 2008) and the 2022 Russia-Ukraine war (**123 dollar per barrel** in March 2022), 
as well as extreme low-price months during the 2020 COVID-19 demand collapse (**16–17dollar per barrel**). 
These are real, historically documented events — not data entry errors. We retain all 
outliers in the cleaned dataset, as they represent genuine market extremes that are meaningful 
for analysis. We flag them with an is_outlier column for optional use downstream.

### Flag Outliers

```python
df_clean['is_outlier'] = (df_clean['price'] < lower) | (df_clean['price'] > upper)
print(f"Outlier flag added. {df_clean['is_outlier'].sum()} rows flagged.")
df_clean.head()
```

### Dataset #3: Brazil Weather — INMET Station Network
**Source:** https://www.kaggle.com/datasets/gregoryoliveira/brazil-weather-information-by-inmet  
**Raw shape:** 3,289,529 station-day rows across 647 stations  
**Final shape:** 309 monthly observations after aggregation  
**Variables used:** `temp_mean` (°C), `precip_mm` (mm/day)

Brazil produces over a third of the world's coffee, making it the single most important country for global supply. Climate shocks with their drought and heat stress seasons affect coffee production during the flowering and bean development stages. These are well-documented drivers of supply shortfalls. The 2021 drought, for example, contributed to one of Brazil's worst harvests in 20 years and is cited in our background as a key source of recent price increases.

Raw observations are at the station-day level across 647 INMET automated stations. Of the 11 original columns, only mean daily temperature and daily precipitation were retained. This is the actual variable entering the regression.

The main limitation to understand is the geography: coffee production in Brazil is concentrated in Minas Gerais, São Paulo, and Espírito Santo, but the 647 stations span the entire country. ### Dataset #4: Vietnam Daily Weather by Province
**Source:** https://www.kaggle.com/datasets/vanviethieuanh/vietnam-weather-data  
**Raw shape:** 181,960 province-day rows across 63 provinces  
**Final shape:** 150 monthly observations after aggregation  
**Variables used:** `temp_mean` (°C, derived from daily high/low), `precip_mm` (mm/day)

Vietnam is the world's second largest coffee producer and the dominant supplier of Robusta beans. Its climate — particularly monsoon-driven rainfall in the Central Highlands — directly affects yield timing and volume. Including Vietnam alongside Brazil allows the regression to capture supply-side climate shocks across both major production regions simultaneously, consistent with the 6–12 month lag structure described in our hypothesis.

Raw observations are at the province-day level across 63 provinces. Mean daily temperature is not provided directly and is derived as the average of the daily high and low. No missing values are present in the raw data. The dataset ends in June 2021, cutting off the final six months of our intended 2009–2021 window — this is noted as a limitation but does not substantially affect the lagged regression structure.

Spatial aggregation bias is a limitation here as it is for Brazil, but the mechanics differ. 

Brazil's bias stems from sheer land area as the 647 stations are spread across 8.5 million km². Vietnam is far smaller at roughly 330,000 km², but the country's elongated north-to-south shape produces dramatic climate variation between regions. The north is cooler and more seasonally distinct, while the south is tropical year-round. 

Coffee production is concentrated almost entirely in two Central Highlands provinces, Đắk Lắk and Lâm Đồng. Their highland elevation and monsoon patterns are climatically distinct from the national average. The average across all 63 provinces folds in Hanoi's climate, Ho Chi Minh City's climate, and coastal provinces that have no meaningful relationship to coffee yield outcomes. 

Unlike Brazil, where station-level coordinates allow the coverage gap to be visualized directly, the Vietnam dataset provides no station-level metadata, so this bias cannot be illustrated spatially and is instead acknowledged here as a constraint on causal interpretation.

### Combining Brazil & Vietnam

Both country-level monthly tables are merged on `year_month` using an outer join. 

The outer join preserves Brazil's full range (May 2000–Jan 2026) for potential future use, while Vietnam contributes only its available window (Jan 2009–Jun 2021). The 159 NaN values in the Vietnam columns are structurally missing since they correspond to months outside Vietnam's data range; these are not measurement failures.

For the regression analysis, the combined dataset will be scoped to the 2009–2021 overlap period where both countries have complete observations. This is consistent with the date range specified in our research question. Climate anomalies, which are deviations from each country's long-run monthly mean, will be computed from these cleaned averages during the EDA phase.

```python
climate_monthly = brazil_monthly.rename(columns={
    'temp_mean': 'brazil_temp_mean',
    'precip_mean': 'brazil_precip_mean'
})

climate_monthly = climate_monthly.merge(
    vietnam_monthly.rename(columns={
        'temp_mean': 'vietnam_temp_mean',
        'precip_mean': 'vietnam_precip_mean'
    }),
    on='year_month',
    how='outer'
)

climate_monthly.sort_values('year_month', inplace=True)
climate_monthly.reset_index(drop=True, inplace=True)

print("Combined shape:", climate_monthly.shape)
print(f"Date range: {climate_monthly['year_month'].min()} to {climate_monthly['year_month'].max()}")
print("\nMissingness (Vietnam NaNs = outside 2009–2021 coverage, not gaps):")
print(climate_monthly.isnull().sum())

climate_monthly
```
